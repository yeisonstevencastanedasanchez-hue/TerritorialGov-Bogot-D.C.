```python
"""
Ejemplo práctico sencillo de LlamaIndex - OpenSource
Este ejemplo muestra cómo crear un índice simple y hacer consultas
"""

from llama_index.core import Document, VectorStoreIndex
from llama_index.core.embeddings import MockEmbedding
from llama_index.core.llms import MockLLM
from llama_index.core.settings import Settings

# Usar embedding mock y LLM mock (completamente local y sin necesidad de API keys)
Settings.embed_model = MockEmbedding(embed_dim=384)
Settings.llm = MockLLM(max_tokens=256)

# 1. Crear documentos de ejemplo
documentos = [
    Document(text="Python es un lenguaje de programación versátil y fácil de aprender."),
    Document(text="LlamaIndex es una librería para construir aplicaciones con LLMs."),
    Document(text="Los datos estructurados mejoran la calidad de las respuestas de IA."),
    Document(text="El aprendizaje automático requiere datos limpios y bien etiquetados."),
]

# 2. Crear un índice de vectores
print("Creando índice de vectores...")
indice = VectorStoreIndex.from_documents(documentos)
print("✓ Índice creado exitosamente\n")

# 3. Crear un motor de consultas
motor_consulta = indice.as_query_engine()

# 4. Hacer consultas al índice
consultas = [
    "¿Qué es LlamaIndex?",
    "¿Por qué son importantes los datos?",
    "¿Cuál es el lenguaje Python?",
]

print("=" * 50)
print("CONSULTAS AL ÍNDICE")
print("=" * 50)

for consulta in consultas:
    print(f"\n📝 Consulta: {consulta}")
    respuesta = motor_consulta.query(consulta)
    print(f"💡 Respuesta: {respuesta}")
    print("-" * 50)
```
