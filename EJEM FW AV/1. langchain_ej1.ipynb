```python
# EJEMPLO 2: Usando Prompts y el modelo directamente (Método Moderno)

# En versiones modernas de LangChain, los prompts se usan directamente con el modelo
from langchain_core.prompts import PromptTemplate  # Plantilla para prompts
from langchain_ollama import OllamaLLM  # Modelo

# Creamos un template de prompt
# Las llaves {} son variables que sustituiremos después
template_prompt = """Analiza el siguiente texto y proporciona un resumen en una oración:
Texto: {texto}
Resumen:"""

# Creamos un objeto PromptTemplate
# input_variables lista las variables que usaremos en el template
prompt = PromptTemplate(
    input_variables=["texto"],  # Variable que irá donde está {texto}
    template=template_prompt  # El template que creamos
)

# Usamos OllamaLLM
llm = OllamaLLM(model="mistral", base_url="http://localhost:11434")

# Creamos una cadena (pipe) moderno: prompt -> modelo
# El símbolo | conecta el prompt con el modelo
cadena = prompt | llm

# Ejecutamos la cadena con un texto de entrada
texto_entrada = "Python es un lenguaje de programación versátil usado para ciencia de datos, desarrollo web y automatización"

# Invocamos la cadena con el texto
resultado = cadena.invoke({"texto": texto_entrada})

# Mostramos el resultado
print("Resumen generado:")
print(resultado)
```
